## Part II: Python Implementation (15 minutes)

Let's use `scikit-learn` to build a decision tree
and a random forest.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

### Step 1: Load the data

Create a small loan default dataset and split
into training and test sets.

```python
data = {
    'Income': [50, 80, 120, 40, 60, 200, 30, 90, 150, 70],
    'Credit_Score': [550, 700, 750, 500, 650, 800, 600, 720, 780, 680],
    'Default': [1, 0, 0, 1, 1, 0, 1, 0, 0, 1]
}
df = pd.DataFrame(data)
df.head()
```

### Step 1b: Prepare features and split

Separate features (X) and target (y), then create
training and test sets.

```python
X = df[['Income', 'Credit_Score']]
y = df['Default']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

### Step 2: Fit a single Decision Tree

TODO: Fit a `DecisionTreeClassifier` with `max_depth=2`
and calculate the test accuracy.

```python
tree = DecisionTreeClassifier(max_depth=2)
tree.fit(X_train, y_train)

tree_pred = tree.predict(X_test)
tree_acc = accuracy_score(y_test, tree_pred)

print(f"Decision Tree Test Accuracy: {tree_acc:.2f}")
```

### Step 3: Fit a Random Forest

TODO: Fit a `RandomForestClassifier` with
`n_estimators=100` and `max_depth=2`,
then calculate the test accuracy.

```python
forest = RandomForestClassifier(n_estimators=100, max_depth=2)
forest.fit(X_train, y_train)

forest_pred = forest.predict(X_test)
forest_acc = accuracy_score(y_test, forest_pred)

print(f"Random Forest Test Accuracy: {forest_acc:.2f}")
```
